# Fase 2: Limpieza y Consolidación

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StringType
from pyspark.sql import functions as F
```

```python
# Create the spark session
spark = SparkSession.builder \
    .appName("Bronze") \
    .config("spark.sql.session.timeZone", "UTC") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")
```

```python
# Read the data from each dir

df_source_1 = spark.read \
    .option("multiline", "true") \
    .json("/app/data/bronze/data_source_1") \
    .withColumn("user_id", F.lit("source_1"))

df_source_2 = spark.read \
    .option("multiline", "true") \
    .json("/app/data/bronze/data_source_2") \
    .withColumn("user_id", F.lit("source_2"))
```

```python
print("--- Numero de registros en cada df ---")
print(f"Origen 1: {df_source_1.count():,}")
print(f"Origen 2: {df_source_2.count():,}")
```

```python
# Consolidate in a single DataFrame
df_consolidated = df_source_1.unionByName(df_source_2)
```

```python
print(f"Numero de registros totales: {df_consolidated.count():,}")
```

```python
# Output the data as parquet
parquet_output_path = "../data/bronze/consolidated_raw.parquet"

df_consolidated.write \
    .mode("overwrite") \
    .parquet(parquet_output_path)
```

```python
df_raw = spark.read.parquet("/app/data/bronze/consolidated_raw.parquet")
```

```python
df_raw.printSchema()
```

```python
# Number of total registers

total_ingested = df_raw.count()

print(f"--- Numero de registros totales: {total_ingested:,} ---")
```

```python
# Drop the "ip_addr" column for data protection issues
df_raw = df_raw.drop("ip_addr")
```

```python
# Show for the first 5 registers
df_raw.show(5, vertical=True)
```

```python
# Count the original number of nulls
print("--- Conteo de valores NULL nativos por columna ---")
df_raw.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c) 
    for c in df_raw.columns
]).show(vertical=True)
```

```python
# Cleaning nulls
garbage_text_values = ["nan", "null", "na", "", "none", " "]
string_columns = [field.name for field in df_raw.schema if isinstance(field.dataType, StringType)]

df_cleaned = df_raw
for col_name in string_columns:
    df_cleaned = df_cleaned.withColumn(
        col_name,
        F.when(
            F.lower(F.trim(F.col(col_name))).isin(garbage_text_values) | F.col(col_name).isNull(),
            F.lit(None).cast(StringType())
        ).otherwise(F.col(col_name))
    )

df_cleaned = df_cleaned.dropna(subset=["ts", "ms_played"])

df_cleaned = df_cleaned.filter((F.col("ms_played").isNotNull()) & (F.col("ms_played") > 0))

null_fill_blueprint = {
    "master_metadata_track_name": "Unknown Track",
    "master_metadata_album_artist_name": "Unknown Artist",
    "master_metadata_album_album_name": "Unknown Album",
    "platform": "Unknown Platform",
    "conn_country": "Unknown Country",
    "episode_name": "Not A Podcast",
    "episode_show_name": "Not A Podcast",
    "spotify_episode_uri": "Not A Podcast",
    "reason_start": "unknown",
    "reason_end": "unknown",
    "offline_timestamp": 0
}

print(f"--- Numero de valores nulos dentro del dataset ---")
df_cleaned.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c) 
    for c in df_cleaned.columns
]).show(vertical=True)

df_cleaned = df_cleaned.fillna(null_fill_blueprint)
```

```python
# Remove duplicated values
total_duplicates = total_ingested - df_cleaned.dropDuplicates(["ts", "user_id"]).count()
print(f"Total de registros duplicados detectados: {total_duplicates:,}")

df_cleaned = df_cleaned.dropDuplicates(["ts", "user_id"])
```

```python
# Add dates, year, month, day of the week, hour, and minutes columns to the df
df_cleaned = df_cleaned \
    .withColumn("date_timestamp", F.to_timestamp("ts")) \
    .withColumn("readable_date", F.to_date("date_timestamp")) \
    .withColumn("year", F.year("date_timestamp")) \
    .withColumn("month", F.month("date_timestamp")) \
    .withColumn("day_of_week", F.dayofweek("date_timestamp")) \
    .withColumn("hour", F.hour("date_timestamp")) \
    .withColumn("min_played", F.round(F.col("ms_played") / 60000, 2))
```

```python
# Normalize song name, artist, and album name
text_targets = ["master_metadata_track_name", "master_metadata_album_artist_name", "master_metadata_album_album_name"]
for text_col in text_targets:
    df_cleaned = df_cleaned.withColumn(
        text_col,
        F.trim(F.regexp_replace(F.col(text_col), r"[\r\n\t]", " "))
    )
```

```python
# Save cleaned DataFrame to parquet
df_cleaned.write.mode("overwrite").parquet("/app/data/silver/spotify_clean.parquet")

# Export to csv
df_cleaned.coalesce(1).write \
    .mode("overwrite") \
    .option("header", "true") \
    .option("encoding", "UTF-8") \
    .csv("/app/data/silver/spotify_clean_delivery")
```

```python
print(f"Numero de registros despues de la limpieza: {df_cleaned.count():,}")
```

```python
# Display of the data after being cleaned
df_cleaned.show(5, vertical=True)
```

# Fase 3: Análisis Exploratorio (EDA)

```python
df = spark.read.parquet("/app/data/silver/spotify_clean.parquet")
```

## Analisis 1: Patrones Temporales

```python
# Count of artists and songs
artists_count = df.select("master_metadata_album_artist_name").distinct().count()
songs_count = df.select("spotify_track_uri").distinct().count()

print(f"Total de artistas unicos escuchados: {artists_count:,}")
print(f"Total de canciones unicas escuchadas: {songs_count:,}")

# Most listened artist
top_artist = df.filter(F.col("master_metadata_album_artist_name") != "Unknown Artist") \
    .groupBy("master_metadata_album_artist_name") \
    .count() \
    .orderBy(F.col("count").desc()) \
    .first()

print(f"Artista más escuchado: {top_artist['master_metadata_album_artist_name']} ({top_artist['count']:,} reproducciones)")

# Most listened song
top_song = df.filter(F.col("master_metadata_track_name") != "Unknown Track") \
    .groupBy("master_metadata_track_name", "master_metadata_album_artist_name") \
    .count() \
    .orderBy(F.col("count").desc()) \
    .first()

print(f"Canción más escuchada: '{top_song['master_metadata_track_name']}' de {top_song['master_metadata_album_artist_name']} ({top_song['count']:,} reproducciones)")
```

### Distribución de escucha por hora del día

```python
df_hour = df.groupBy("hour").count().orderBy("hour").toPandas()

day_labels = {1: "Dom", 2: "Lun", 3: "Mar", 4: "Mié", 5: "Jue", 6: "Vie", 7: "Sáb"}
df_day = df.groupBy("day_of_week").count().orderBy("day_of_week").toPandas()
df_day["day_name"] = df_day["day_of_week"].map(day_labels)

plt.figure()
plt.bar(df_hour["hour"], df_hour["count"])
plt.title("Distribución de Escucha por Hora del Día")
plt.xlabel("Hora (Format de 24h)")
plt.ylabel("Reproducciones")
plt.show()

plt.figure()
plt.bar(df_day["day_name"], df_day["count"])
plt.title("Actividad de Escucha por Día de la Semana")
plt.xlabel("Día")
plt.ylabel("Reproducciones")
plt.show()
```

### Tendencia mensual de consumo

```python
df_trend = df.groupBy("month").count().orderBy("month").toPandas()

df_trend["month_axis"] = df_trend["month"].astype(str).str.zfill(2)

plt.plot(df_trend["month_axis"], df_trend["count"])
plt.title("Tendencia Mensual Histórica de Consumo (2015 - 2026)")
plt.xlabel("Mes")
plt.ylabel("Reproducciones")
plt.tight_layout()
plt.show()
```

### Mapa de calor día x hora

```python
df_pivot = df.groupBy("day_of_week").pivot("hour").count().orderBy("day_of_week").toPandas()

df_pivot = df_pivot.set_index("day_of_week")
df_pivot.index = df_pivot.index.map(day_labels)
df_pivot = df_pivot.fillna(0)

sns.heatmap(df_pivot, cmap="YlGnBu", annot=False, fmt="g", cbar_kws={'label': 'Cantidad de Canciones'})
plt.title("Mapa de Calor de Consumo: Día de la Semana vs Hora del Día")
plt.xlabel("Hora del Día")
plt.xticks(rotation=75)
plt.ylabel("Día de la Semana")
plt.yticks(rotation=25)
plt.show()

df_peak = df.groupBy("day_of_week", "hour").count().orderBy(F.col("count").desc()).first()
peak_day = day_labels[df_peak["day_of_week"]]

print("--- Identificacion de hora pico ---")
print(f"Tu momento pico de escucha es los: {peak_day} a las {df_peak['hour']}:00 hrs con {df_peak['count']:,} reproducciones acumuladas.")
```

### Top 20 artistas y canciones más reproducidas

```python
# Top artists
df_top_artists = df.filter(F.col("master_metadata_album_artist_name") != "Unknown Artist") \
    .groupBy("master_metadata_album_artist_name") \
    .count() \
    .orderBy(F.col("count").desc()) \
    .limit(20) \
    .toPandas()

plt.barh(df_top_artists["master_metadata_album_artist_name"], df_top_artists["count"])
plt.gca().invert_yaxis()
plt.title("Top 20 Artistas Más Escuchados")
plt.xlabel("Número de Reproducciones")
plt.ylabel("Artista")
plt.tight_layout()
plt.show()
```

```python
# Top songs
df_top_songs = df.filter(F.col("master_metadata_track_name") != "Unknown Track") \
    .groupBy("master_metadata_track_name", "master_metadata_album_artist_name") \
    .count() \
    .orderBy(F.col("count").desc()) \
    .limit(20) \
    .toPandas()

df_top_songs["song_artist"] = df_top_songs["master_metadata_track_name"] + " (" + df_top_songs["master_metadata_album_artist_name"] + ")"

plt.figure(figsize=(15, 8))
plt.barh(df_top_songs["song_artist"], df_top_songs["count"])
plt.gca().invert_yaxis()
plt.title("Top 20 Canciones Más Escuchada")
plt.xlabel("Número de Reproducciones")
plt.ylabel("Canción (Artista)")
plt.tight_layout()
plt.show()
```

###  Tasa de completación vs skip

```python
df_skip = df.groupBy("skipped").count().toPandas()

df_skip["state"] = df_skip["skipped"].map({True: "skipped", False: "completed_song"})

plt.bar(df_skip["state"], df_skip["count"])
plt.title("Tasa de Completación vs. Skip Global")
plt.xlabel("Total de Eventos")
plt.ylabel("")
plt.tight_layout()
plt.show()
```
